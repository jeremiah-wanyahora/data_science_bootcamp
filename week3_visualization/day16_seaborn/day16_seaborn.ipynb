```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\jerew\Desktop\github_projects\data_science_bootcamp\week2_lessons\day14_project\cleaned_messy_sales.csv")

df.head()
```

```python
# Histplot of revenue

sns.histplot(df[" Revenue"])

plt.title("Distribution of sales revenue")

plt.xlabel("Revenue")

plt.ylabel("Frequency")

plt.figure(figsize=(10, 6))

plt.savefig("revenue_histogram.png")

plt.show()
```

```python
# Boxplot(outlier detection)

sns.boxplot(x=df[" Revenue"])

plt.title("Boxplot of sales revenue")

plt.xlabel("Revenue")

plt.figure(figsize=(10, 5))

plt.savefig("boxplot_of_revenue.png")

plt.show()
```

```python
# Scatterplot of quantity vs revenue

sns.scatterplot(
    x= df[" Quantity"],
    y= df[" Revenue"]
)

plt.title("Scatterplot of Quantity vs Revenue")

plt.xlabel("Quantity")

plt.ylabel("Revenue")

plt.figure(figsize=(10, 6))

plt.savefig("quantity_vs_revenue_scatterplot.png")

plt.show()
```

```python
# Adding colour using hue to map out how regions behave in a scatterplot

sns.scatterplot(
    x= df[" Quantity"],
    y= df[" Revenue"],
    hue= df[" Region"]
)

plt.title("Quantity vs Revenue by Region")

plt.xlabel("Quantity")

plt.ylabel("Revenue")

plt.figure(figsize=(10, 6))

plt.savefig("quantity_vs_revenue_scatterplot_by_region.png")

plt.show()
```

```python
# Boxplot of revenue distributions across regions

sns.boxplot(
    x= df[" Region"],
    y= df[" Revenue"]
)

plt.title("Revenue distribution across regions")

plt.xlabel("Region")

plt.ylabel("Revenue")

plt.figure(figsize=(10, 6))

plt.savefig("revenue_across_regions.png")

plt.show()
```
