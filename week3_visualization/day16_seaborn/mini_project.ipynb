```python
# Create a:
# Histogram of revenue distribution
#Scatterplot of quantity vs revenue
#Boxplot of revenue by product
#Scatterplot of quantity vs revenue coloured by product

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\jerew\Desktop\github_projects\data_science_bootcamp\week2_lessons\day14_project\cleaned_messy_sales.csv")

df.head()

print(df)
```

```python
# Histogram of revenue distribution

sns.histplot(df[" Revenue"])

plt.title(" Revenue Distribution")

plt.xlabel("Revenue")

plt.ylabel("Frequency")

plt.figure(figsize=(10, 5))

plt.savefig("revenue_distribution.png")

plt.show()
```

```python
# Scatterplot of quantity vs revenue

sns.scatterplot(
    x= df[" Quantity"],
    y= df[" Revenue"]
)

plt.title("Quantity vs Revenue")

plt.xlabel("Quantity")

plt.ylabel("Revenue")

plt.figure(figsize=(10,6))

plt.savefig("quantity_vs_revenue_scatterplot.png")

plt.show()
```

```python
#Boxplot of revenue by product

sns.boxplot(
    x= df[" Product"],
    y= df[" Revenue"]
)

plt.title("Revenue by product")

plt.xlabel("Product")

plt.ylabel("Revenue")

plt.figure(figsize=(10, 6))

plt.savefig("revenue_by_product_boxplot.png")

plt.show()
```

```python
#Scatterplot of quantity vs revenue coloured by product

sns.scatterplot(
    x= df[" Quantity"],
    y= df[" Revenue"],
    hue= df[" Product"]
)

plt.title("Quantity vs Revenue by Product")

plt.xlabel("Quantity")

plt.ylabel("Revenue")

plt.figure(figsize=(10, 6))

plt.savefig("quantity_vs_revenue_by_product.png")

plt.show()
```
