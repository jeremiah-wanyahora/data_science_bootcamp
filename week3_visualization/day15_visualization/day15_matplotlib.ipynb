```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\jerew\Desktop\github_projects\data_science_bootcamp\week2_lessons\day14_project\cleaned_messy_sales.csv")

print(df)

df.head()
```

```python
# Visualize the total sales by product category

sales_by_product_category = df.groupby(" Product")[" Revenue"].sum()

sales_by_product_category.plot(kind="bar", title="Total Sales by Product Category", ylabel="Total Sales", xlabel="Product Category", figsize=(10, 6))

plt.savefig("sales_by_product_category.png")

plt.show()
```

```python
# Monthly sales  trend by line 

sales_by_month = df.groupby("Month")[" Revenue"].sum()

sales_by_month.plot(kind="line")

plt.title("Monthly Sales Trend")

plt.xlabel("Month")

plt.ylabel("Total Sales")

plt.savefig("monthly_sales_trend.png")

plt.show()

```

```python
# Create a histogram of sales distribution

df[" Revenue"].plot(kind="hist")

plt.title("Distribution of Sales")

plt.xlabel("Sales Amount")

plt.ylabel("Frequency")

plt.figure(figsize=(10,5))

plt.savefig("sales_distribution.png")

plt.show()
```
