```python
# Visualize a:
# Bar chart - revenue by region
#Line chart - sales trend by month
# Histogram - quantity distribution

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\jerew\Desktop\github_projects\data_science_bootcamp\week2_lessons\day14_project\cleaned_messy_sales.csv")

print(df)
```

```python
revenue_by_region = df.groupby(" Region")[" Revenue"].sum()

revenue_by_region.plot(kind="bar")

plt.title("Revenue by region")

plt.xlabel("Region")

plt.ylabel("Revenue")

plt.figure(figsize=(10,6))

plt.savefig("revenue_by_region.png")

plt.show()
```

```python
# Line chart - sales trend by month

sales_trend = df.groupby("Month")[" Revenue"].sum()

sales_trend.plot(kind="line")

plt.title("Sales trend by month")

plt.xlabel("Month")

plt.ylabel("Revenue")

plt.figure(figsize=(10,5))

plt.savefig("sales_trend_by_month.png")

plt.show()
```

```python
# Histogram - quantity distribution 

df[" Quantity"].plot(kind="hist", bins=20)

plt.title("Quantity distribution")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.figure(figsize=(10,5))
plt.savefig("quantity_distribution.png")
plt.show()
```
