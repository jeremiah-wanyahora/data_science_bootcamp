```python
import pandas as pd

df= pd.read_csv("messy_sales.csv", on_bad_lines='skip')

df.head()

print(df)
```

```python
import numpy as np

df = df.replace(r'^\s*$', np.nan, regex=True)

print(df)
```

```python
print(df.info())
```

```python
#Check for missing values

df.isnull().sum()

#You will discover: missing quantity(2), missing Revenue entry(1) and missing customer names(2)
```

```python
df.duplicated().sum()
```

```python
#Standardize product names

df[" Product"] = df[" Product"].str.strip().str.title()

#Standardize Region names

df[" Region"] = df[" Region"].str.strip().str.title()

#Standardize Customer names

df[" Customer"] = df[" Customer"].str.strip().str.title()

print(df)

```

```python
#Convert Quantity to numeric

df[" Quantity"] = pd.to_numeric(df[" Quantity"], errors= "coerce")

#Convert Revenue to numeric

df[" Revenue"] = pd.to_numeric(df[" Revenue"], errors= "coerce")

#Convert Date from object to datetime

df["Date"] = pd.to_datetime(df["Date"], errors= "coerce")

print(df)
```

```python
# Handle missing values

#Remove invalid rows

df = df.dropna(subset=["Date", " Revenue", " Quantity", " Customer"])

print(df)
```

```python
#Remove negative values

df = df[(df[" Quantity"] >= 0) & (df[" Revenue"] >= 0)]

df = df.drop_duplicates()
```

```python
#Data analysis and creation of new columns to mark trends 

#Create a new column for price per unit

df["Price"] = df[" Revenue"] / df[" Quantity"]


#Create a new column to extract month from date

df["Month"] = df["Date"].dt.month

print(df)

#
```

```python
# Categorize sales by product

sales_by_product = df.groupby(" Product")[" Revenue"].sum()

print(sales_by_product)

sales_by_region = df.groupby(" Region")[" Revenue"].sum()
print(sales_by_region)
```

```python
df.to_csv("cleaned_messy_sales.csv", index=False)
```
