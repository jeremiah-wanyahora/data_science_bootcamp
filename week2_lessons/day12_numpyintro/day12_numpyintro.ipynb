```python
# Import numpy and create arrays

import numpy as np

# Create a 1D array(vector)

sales = np.array([200, 450, 300, 500, 800])

# Create a 2D array (matrix)

costs = np.array([
    [100, 220],
    [200, 180],
    [150, 140]
])
```

```python
# Perform vectorized operations (No loops)

profits = sales - 0.3 * sales
print("After tax profits:", profits)
```

```python
# Array slicing

print("first sale:", sales[0]) # Returns the first sale
print("second sale", sales[1]) # Returns second sale entry
print("last sale:", sales[-1]) # Returns last sale
print("second last sale:", sales[-2]) # Returns second last sale
print("2nd to 4th sales:", sales[1:4]) # Returns second to forth sales
```

```python
# Array slicing
print(costs[1:3])
print(costs[0])
print(costs[-1])
```

```python
# Numpy functions

print("Average Sales:", np.mean(sales))
print("Highest Sale:", np.max(sales))
print("Lowest Sale:", np.min(sales))
print("Standard deviation:", np.std(sales))
print("Total Sales:", sum(sales))
```

```python
# Reshaping and shape checking


print(costs.shape)
sales_reshaped = sales.reshape(5, 1)
print(sales_reshaped)
costs_reshaped = costs.reshape(2,3)
print(costs_reshaped)
print(sales_reshaped.shape)
print(sales_reshaped)
print(sales.shape)
print(sales)
```

```python
# Broadcasting; Add bonus of 50 to all sales

bonus_sales = sales + 50
print(bonus_sales)
```

```python
# Comparison and filtering

# Categorize high sales as those that are above 400

high_sales = sales[sales > 400]
print("High record sales:", high_sales)

# Categorize sales below 500 as low selling

low_sales = bonus_sales[bonus_sales < 500]
print("Low selling products:", low_sales)
```

```python
# Create a numpy array 

import numpy as np

monthly_expenses = np.array([12000, 11000, 10500, 13000, 9500])

print("Monthly expenses:", monthly_expenses)

# Calculate average and standard deviation.

print("Average Monthly expenses:", np.mean(monthly_expenses))
print("Standard deviation:", np.std(monthly_expenses))
```

```python
# Subtract each value from a monthly budget (say KSh 15,000)

Disposable_income = 15000 - monthly_expenses
print("Money left every month:", Disposable_income)
```

```python
# Find which months were under budget.

underbudget_months = monthly_expenses[monthly_expenses > 15000]
print("Under budget months:", underbudget_months)
```

```python
# Reshape the data into a column format.

monthly_expenses_reshaped = monthly_expenses.reshape(5,1)
print(monthly_expenses_reshaped)
```
