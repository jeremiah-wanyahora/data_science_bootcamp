```python
import pandas as pd

# Load csv

df = pd.read_csv(r"C:\Users\Jeremiah Muraguri\Desktop\data_science_bootcamp\week2_lessons\day11_aggregation\clean_sales.csv")

print(df)
```

```python
# Standardize product names

df["Product"] = df["Product"].str.title().str.strip()

# Change date from object dtype to datetime

df["Date"] = pd.to_datetime(df["Date"])



# Create new column Bigorder

df["Bigorder"] = df["Quantity"] > 10

# Create Revenue column

df["Revenue"] = df["Price"] * df["Quantity"]

# Create Profit column

df["Profit"] = df["Revenue"] - (df["Cost"] * df["Quantity"])

# Create margin column

df["Profit_margin"] = (df["Profit"] / df["Revenue"]) * 100

# Create new column High_margin

df["High_margin"] = df["Profit_margin"] > 35

print(df.dtypes)
print(df)
```

```python
# Save the file 

df.to_csv("enhanced_sales.csv", index=False)
```
