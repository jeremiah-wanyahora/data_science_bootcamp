```python
import pandas as pd

df = pd.read_csv("sales.csv")

# Practice tasks
# Print all rows where: 
# Region is not Nairobi 

df[df["Region"] != "Nairobi"]
```

```python
# Quantity is greater than 10

df[df["Quantity"] > 10]
```

```python
# Show all products sold in Mombasa with a price over 1000

df[(df["Region"] == "Mombasa") & (df["Price"] > 1000)]
```

```python
# Extract only ["Product", "Price", "Revenue"] for the first 5 rows

df.loc[0:6, ["Product", "Price", "Region"]]
```

```python
# BONUS: Create a new column Profit = Revenue - (Cost * Quantity), and show top 3 profitable sales

# Create new column Revenue

df["Revenue"] = df["Price"] * df["Quantity"]

# Create another new column Profit

df["Profit"] = df["Revenue"] - (df["Cost"] * df["Quantity"])

df.sort_values("Profit", ascending = False).head(3)
```
