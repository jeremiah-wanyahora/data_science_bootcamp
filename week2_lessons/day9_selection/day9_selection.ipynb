```python
# Accessing columns

import pandas as pd

# Load csv
df = pd.read_csv("sales.csv")

print("Products")
df["Product"] # View single column

df[["Product", "Price"]] # View multiple columns

```

```python
# Accessing rows by position

df.iloc[0]   # Returns first row
df.iloc[:3]  # Returns first 3 rows
df.iloc[1:4]   # Returns row 1-3(excludes row 4)
df.iloc[2:6]   # Returns row 2-5(excludes row 6)
```

```python
# Accessing label

df.loc[2, "Product"]
df.loc[0:2, ["Product", "Region"]]
```

```python
# Region and price for the first 3 records

df.loc[0:2, ["Region", "Price"]]
```

```python
# Boolean filtering(Conditionals)

# All Mombasa orders

df[df["Region"] == "Mombasa"]

# Cheap products

df[df["Price"] < 600 ]

# High ticket items sold in Nairobi

df[(df["Region"] == "Nairobi") & (df["Price"] > 600)]

# Products not sold in kisumu

df[df["Region"] != "Kisumu"]
```

```python
# Combine Conditions with & and |

# Coffee or tea sold in Nairobi

df[(df["Product"].isin(["Coffee", "Tea"])) & (df["Region"] == "Nairobi")]
```

```python
# Using isin and notna

df[df["Product"].isin(["Coffee", "Tea"])]

# Drop rows with missing values

df[df["Price"].notna()]
```
