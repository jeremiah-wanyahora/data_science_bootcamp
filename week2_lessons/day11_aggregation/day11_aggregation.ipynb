```python
import pandas as pd

df = pd.read_csv("sales_raw.csv")
print(df)
```

```python
# Data Cleaning: Changing the entry marked as unknown in region

df.loc[(df["Product"] == "Tea") & (df["Quantity"] == 20.0), "Region"] = "Mombasa"
print(df)
```

```python
df.duplicated().sum()
```

```python
df_cleansales = df.drop_duplicates()
print(df_cleansales)
```

```python
df_cleansales.to_csv("clean_sales.csv", index=False)
```

```python
# Data aggregation 
# Load clean csv

ds = pd.read_csv("clean_sales.csv")
print(ds)
```

```python
# Create revenue and profit columns

ds["Revenue"] = ds["Price"] * ds["Quantity"]
ds["Profit"] = ds["Revenue"] - (ds["Cost"]*ds["Quantity"])
print(ds)

# Print a small report

print("Total Revenue: Ksh.", ds["Revenue"].sum())
print("Total Profit: Ksh.", ds["Profit"].sum())
print("Average Price: Ksh.", ds["Price"].mean())
```

```python
# Group by region or product

print("Revenue per product:", ds.groupby("Product")["Revenue"].sum())
print("Revenue per Region:",ds.groupby("Region")["Revenue"].sum())
```

```python
# Multiple aggregations with .agg()

print("Grouping by product", 
    ds.groupby("Product").agg({
    "Revenue": "sum",
    "Price": "mean",
    "Profit": "sum"
}))

print("Grouping by region", 
    ds.groupby("Region").agg({
    "Revenue": "sum",
    "Profit": "sum"
}))
```

```python
# Top 3 products per revenue

print(ds.groupby("Region")["Revenue"].sum().sort_values(ascending=False).head(3))
print(ds.groupby("Product")["Revenue"].sum().sort_values(ascending=False).head(3))
```

```python
# Total revenue per product per region

print(ds.groupby(["Product", "Region"])["Revenue"].sum())
print(ds.groupby(["Region", "Product"]).agg({
    "Revenue": "sum",
    "Profit": "sum",
    "Quantity": "mean"
}))

# Multi indexing lets you segment performance well
```

```python
# Group by Region and show total Revenue and total Profit

regional_summary = ds.groupby("Region").agg({
    "Revenue": "sum",
    "Profit": "sum"
})
```

```python
# For each Product, show average Price, Quantity sold, and total Revenue.

product_summary = ds.groupby("Product").agg({
    "Price": "mean",
    "Quantity": "sum",
    "Revenue": "sum"
})
```

```python
# Which region has the highest average Cost

highest_average_cost = ds.groupby("Region")["Cost"].mean().sort_values(ascending=False).head(1)
```

```python
# Which product generates the highest profit

highest_profit = ds.groupby("Product")["Profit"].sum().sort_values(ascending=False).head(1)
```

```python
# Save the reports

regional_summary.to_csv("regional_summary.csv", index=True)
```

```python
product_summary.to_csv("product_summary.csv", index=True)
```
