```python
# Load CSV

import pandas as pd

df = pd.read_csv("sales_raw.csv") # Load the data
print(df)
```

```python
# Data cleaning
# Missing values

df.isna().sum()
print("Missing entries of Region:", df["Region"].isna().sum())
print("Missing entries of Cost:", df["Cost"].isna().sum())
print("Missing entries of Quantity:", df["Quantity"].isna().sum())
```

```python
# Handle missing values by either droping rows or filling them
# i. Dropping rows

df.dropna() # deletes all rows with missing values
```

```python
# ii. Filling values

df["Region"] = df["Region"].fillna("Unknown")
df["Cost"] = df["Cost"].fillna(0)
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
print(df)
                                
```

```python
# Convert data types

df.dtypes

# Change price from object to float

df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

print(df)
```

```python
# Remove duplicates

print("Number of duplicated row(s):", df.duplicated().sum())
```

```python
# Remove duplicated rows

df = df.drop_duplicates()
print(df)
```

```python
# Save the csv file as cleaned _sales

df.to_csv("cleaned_sales.csv", index=False)
print(df)
```
