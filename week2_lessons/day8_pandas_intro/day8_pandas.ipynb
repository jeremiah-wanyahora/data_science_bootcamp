```python
# Day 8: Introduction to Pandas. Loading and Exploring sales data

# Step 1: Import Pandas

import pandas as pd

# Read csv
df = pd.read_csv("sample_sales.csv")

# Load the first five rows
print("Preview of dataset")
df.head()
```

```python
# Understanding structure and data types

print("Shape of dataset:", df.shape) # rows, columns
print("Column names:", df.columns.tolist())
print("Data Types:")
df.dtypes
```

```python
# Summary statistics

print("Descriptive summary")
df.describe()
```

```python
# Unique regions

print("Regions in dataset:", df["Region"].unique())

# Count of products

print(("Total products:"), df["Product"].value_counts())
```

```python
# Add a new column revenue by calculation

df["Revenue"] = df["Price"] * df["Quantity"]

# Add a new column profit by calculation

df["Profit"] = df["Price"] - df["Cost"]

# Load csv now

df.head()
```

```python
# Group revenue by product

print("Total revenue per product:")
print(df.groupby("Product")["Revenue"].sum())

# Group revenue by region

print("Revenue by region:")
print(df.groupby("Region")["Revenue"].sum())

# Group profit by product

print("Profit per product:")
print(df.groupby("Product")["Profit"].sum())
```

```python
# Filter examples using real world scenarios 

# Only Nairobi orders

nairobi_orders = df[df["Region"] == "Nairobi"]
print("Nairobi orders")
print(nairobi_orders)

# Expensive orders

expensive_orders = df[df["Price"] > 1000]
print("Expensive orders")
print(expensive_orders)

# Coffee orders in Nairobi

coffee_nairobi_orders = df[(df["Product"] == "Coffee") & (df["Region"] == "Nairobi")]
print("Nairobi orders for coffee")
print(coffee_nairobi_orders)
```

```python
# Save the new enriched version

# Save the cleaned/enriched version

df.to_csv("cleaned_sales.csv", index=False)
print(" Cleaned file saved as 'cleaned_sales.csv'")
```
